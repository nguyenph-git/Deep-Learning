#### (a) Exponential Linear Unit (ELU), $\alpha = 0.1$

$$\text{ELU}(x)=\begin{cases}0.1\left(\exp(x)-1\right) & \text{if } x\leq 0\\ x & \text{if } x>0\end{cases}$$

**Output range:** $(-0.1, \infty)$. For $x>0$ the output is $x$ (unbounded above); for $x\leq 0$ it is $0.1(\exp(x)-1)$, which decreases towards $-0.1$ as $x\to-\infty$ but never reaches it.

**Derivative (step by step):**

- Left branch ($x < 0$): $\dfrac{d}{dx}\big(0.1\exp(x) - 0.1\big) = 0.1\exp(x)$
- Right branch ($x > 0$): $\dfrac{d}{dx}(x) = 1$

$$\text{ELU}'(x)=\begin{cases}0.1\exp(x) & \text{if } x< 0\\ 1 & \text{if } x>0\end{cases}$$

From the left branch, as $x$ moves from $-\infty$ to $0$ the derivative moves from $0$ to $0.1$, since $0.1\exp(-\infty) = 0.1 \times 0 = 0$ and $0.1\exp(0) = 0.1$. From the right branch the derivative is $1$ for all $x > 0$.

Thus the derivative is **not defined at $x = 0$**, because the left-hand limit ($0.1$) does not match the right-hand limit ($1$). Therefore ELU with $\alpha = 0.1$ is differentiable everywhere **except** at $x = 0$. In practice ELU is commonly used with $\alpha = 1$, which makes the two one-sided derivatives agree at $0$ and the function differentiable everywhere.

#### (b) Gaussian Error Linear Unit (GELU)

$$\text{GELU}(x) = x\,\Phi(x)$$

where $\Phi(x)$ is the cumulative distribution function (CDF) of the standard Gaussian $X \sim N(0,1)$, and

$$g(x) = \frac{1}{\sqrt{2\pi}}\exp\!\left(-\frac{x^2}{2}\right)$$

is the corresponding probability density function (PDF).

**Output range:** approximately $[-0.17, \infty)$. For $x>0$, $\text{GELU}(x)=x\Phi(x)>0$ and grows like $x$ (since $\Phi(x)\to 1$), so it is unbounded above. For $x<0$ the output is negative, but $x\Phi(x)\to 0$ as $x\to-\infty$ because $\Phi(x)$ decays faster than $|x|$ grows, so unlike ReLU/ELU the negative part is a small bump rather than a line. The minimum is where $\text{GELU}'(x)=0$, i.e. $\Phi(x)+x\,g(x)=0$, which is solved numerically at $x\approx-0.752$ giving $\text{GELU}(x)\approx-0.170$. Hence the output is bounded below by about $-0.17$ (this is visible in the plot below).

**Derivative (step by step):**

Apply the product rule to $\text{GELU}(x) = x\,\Phi(x)$:

$$\frac{d\,\text{GELU}(x)}{dx} = x\,\Phi'(x) + \Phi(x)$$

By definition, the derivative of a CDF is its PDF:

$$\Phi'(x) = \frac{d\Phi(x)}{dx} = g(x)$$

Substituting back:

$$\frac{d\,\text{GELU}(x)}{dx} = x\,g(x) + \Phi(x) = \frac{x}{\sqrt{2\pi}}\exp\!\left(-\frac{x^2}{2}\right) + \Phi(x)$$

**Practical (tanh) approximation:**

In practice $\text{GELU}(x)$ is expensive to compute, because $\Phi(x)$ is an integral with no elementary closed form. In deep learning libraries GELU is therefore often approximated with a $\tanh$ expression:

$$\text{GELU}(x) \approx 0.5\,x\left(1 + \tanh\!\left(\sqrt{\tfrac{2}{\pi}}\left(x + 0.044715\,x^3\right)\right)\right)$$

which corresponds to approximating the CDF as

$$\Phi(x) \approx 0.5\left(1 + \tanh\!\left(\sqrt{\tfrac{2}{\pi}}\left(x + 0.044715\,x^3\right)\right)\right)$$

Substituting this approximation into the exact derivative gives the form used inside deep learning modules:

$$\frac{d\,\text{GELU}(x)}{dx} \approx \frac{x}{\sqrt{2\pi}}\exp\!\left(-\frac{x^2}{2}\right) + 0.5\left(1 + \tanh\!\left(\sqrt{\tfrac{2}{\pi}}\left(x + 0.044715\,x^3\right)\right)\right)$$


```python
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE" #to run on window machine

import torch
import torch.nn as nn
import matplotlib.pyplot as plt



# requires_grad=True so autograd can give us the derivatives

x = torch.linspace(-5.0, 5.0, 1001, requires_grad=True)

elu  = nn.ELU(alpha=0.1)
gelu = nn.GELU()          # exact (erf) form; nn.GELU(approximate='tanh') for the tanh version

def value_and_grad(module, x):
  """f(x) and f'(x) elementwise. Summing is the trick: d(sum y)/dx_i = dy_i/dx_i."""
  y = module(x)
  grad, = torch.autograd.grad(y.sum(), x, retain_graph=True)
  return y.detach(), grad

y_elu,  d_elu  = value_and_grad(elu,  x)
y_gelu, d_gelu = value_and_grad(gelu, x)
xs = x.detach()

BLUE, ORANGE = "#0072B2", "#D55E00"   # colourblind-safe pair
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2), sharex=True)

panels = [(y_elu, y_gelu, "ELU(x)",  "GELU(x)",  "Activations"),
        (d_elu, d_gelu, "ELU'(x)", "GELU'(x)", "Derivatives")]

for ax, (a, b, la, lb, title) in zip(axes, panels):
  ax.plot(xs, a, lw=2, color=BLUE,   label=la)
  ax.plot(xs, b, lw=2, color=ORANGE, label=lb)
  ax.axhline(0, lw=0.8, color="0.75")
  ax.axvline(0, lw=0.8, color="0.75")
  ax.grid(alpha=0.25, lw=0.6); ax.set_axisbelow(True)
  ax.set_title(title); ax.set_xlabel("x"); ax.legend(frameon=False)
  for s in ("top", "right"):
      ax.spines[s].set_visible(False)

fig.tight_layout()
plt.show()
```

**(a)** You need to develop the class `MyLinear` with the following skeleton. You need to declare the weight matrix and bias of this linear layer.

<div style="text-align: right"><font color="red">[3 points]</font></div>

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

class MyLinear(torch.nn.Module):
  def __init__(self, input_size, output_size):
    """
    input_size: the size of the input
    output_size: the size of the output
    """
    super().__init__()
    #Your code here
    # A linear layer computes  h = x @ W + b  with
    #   W : weight matrix of shape [input_size, output_size]
    #   b : bias vector  of shape [output_size]
    # Both are wrapped in torch.nn.Parameter so that PyTorch registers them as trainable
    # parameters of the module (model.parameters(), model.zero_grad(), state_dict() all work).
    # W is initialised with N(0, 1/input_size) (LeCun initialisation) so that the
    # pre-activations keep a unit scale, b is initialised with zeros.
    self.W = torch.nn.Parameter(torch.randn(input_size, output_size, device=device) / (input_size ** 0.5))
    self.b = torch.nn.Parameter(torch.zeros(output_size, device=device))

  #forward propagation
  def forward(self, x): #x is a mini-batch
    #Your code here
    # x has shape [batch_size, input_size]  ->  output has shape [batch_size, output_size]
    return torch.matmul(x, self.W) + self.b
```

**(b)** You need to develop the class `MyFFN` with the following skeleton

<div style="text-align: right"><font color="red">[7 points]</font></div>

```python
class MyFFN(torch.nn.Module):
  def __init__(self, input_size, num_classes, hidden_sizes, act = torch.nn.ReLU()):
    """
    input_size: the size of the input
    num_classes: the number of classes
    act is the activation function
    hidden_sizes is the list of hidden sizes
    for example input_size = 3, hidden_sizes = [5, 7], num_classes = 4, and act = torch.nn.ReLU()
    means that we are building up a FFN with the confirguration
    (3 (Input) -> 5 (ReLU) -> 7 (ReLU) -> 4 (Output))
    """
    super(MyFFN, self).__init__()
    self.input_size = input_size
    self.num_classes = num_classes
    self.act = act
    self.hidden_sizes = hidden_sizes
    self.num_layers = len(hidden_sizes) + 1

  def _make_act(self):
    """
    `act` can be passed either as a class (torch.nn.ReLU) or as an instance (torch.nn.ReLU()).
    Return a module in both cases.
    """
    return self.act() if isinstance(self.act, type) else self.act

  def create_FFN(self):
    """
    This function creates the feed-forward neural network
    We stack many MyLinear layers
    """
    hidden_sizes = [self.input_size] + self.hidden_sizes + [self.num_classes]
    self.layers = torch.nn.ModuleList()
    #Your code here
    # Hidden layers: each one is a MyLinear followed by the activation function.
    # With hidden_sizes = [784, 100, 100, 10] this builds 784 -> 100 (act) -> 100 (act)
    for index in range(len(self.hidden_sizes)):
        layer = torch.nn.Sequential(MyLinear(hidden_sizes[index], hidden_sizes[index + 1]),
                                    self._make_act())
        self.layers.append(layer)
    # Output layer: a MyLinear WITHOUT activation (100 -> 10). It returns the logits;
    # the softmax is applied inside the cross-entropy loss.
    self.layers.append(MyLinear(hidden_sizes[-2], hidden_sizes[-1]))

  def forward(self, x):
    """
    This implements the forward propagation of the batch x
    This needs to return the logits of x
    """
    result = x
    for layer in self.layers:
        result = layer(result)
    return result

  def compute_loss(self, x, y):
    """
    This function computes the cross-entropy loss
    You can use the built-in CE loss of PyTorch
    """
    #Your code here
    logits = self.forward(x)               # shape [batch_size, num_classes]
    loss_fn = torch.nn.CrossEntropyLoss()  # = softmax + negative log-likelihood, averaged over the batch
    return loss_fn(logits, y)

  # ---- helpers shared by the three optimisers -------------------------------------------
  def _linear_layers(self):
    """All MyLinear layers of the network (the ones that own the parameters W and b)."""
    return [m for m in self.modules() if isinstance(m, MyLinear)]

  def _backward(self, x, y):
    """
    Common first step of every update rule:
    clear the gradients of the previous step, compute the loss on the batch (x, y)
    and back-propagate to obtain dl/dW and dl/db for every layer.
    """
    self.train()
    self.zero_grad()
    loss = self.compute_loss(x, y)
    loss.backward()
    return loss


  def update_SGD(self, x, y, learning_rate = 0.01):
    """
    This function updates the model parameters using SGD using the batch (x,y)
    You need to implement the update rule manually and cannot rely on the built-in optimizer

    Update rule, for every parameter theta with gradient g = dl/dtheta on the mini-batch:
        theta <- theta - eta * g
    """
    #Your code here
    self._backward(x, y)
    with torch.no_grad():
        for layer in self._linear_layers():
            layer.W -= learning_rate * layer.W.grad
            layer.b -= learning_rate * layer.b.grad


  def update_SGDwithMomentum(self, x, y, learning_rate = 0.01, momentum = 0.9):

    #Your code here
    self._backward(x, y)
    with torch.no_grad():
        for layer in self._linear_layers():
            if not hasattr(layer, 'vW'):                 # first call: v = 0
                layer.vW = torch.zeros_like(layer.W)
                layer.vb = torch.zeros_like(layer.b)
            layer.vW.mul_(momentum).add_(layer.W.grad, alpha=1.0 - momentum)
            layer.vb.mul_(momentum).add_(layer.b.grad, alpha=1.0 - momentum)
            layer.W -= learning_rate * layer.vW
            layer.b -= learning_rate * layer.vb


  def update_AdaGrad(self, x, y, learning_rate = 0.01, epsilon = 1e-8):

    #Your code here
    self._backward(x, y)
    with torch.no_grad():
        for layer in self._linear_layers():
            if not hasattr(layer, 'sW'):                 # first call: s = 0
                layer.sW = torch.zeros_like(layer.W)
                layer.sb = torch.zeros_like(layer.b)
            layer.sW.addcmul_(layer.W.grad, layer.W.grad)   # s += g * g
            layer.sb.addcmul_(layer.b.grad, layer.b.grad)
            layer.W -= learning_rate * layer.W.grad / (torch.sqrt(layer.sW) + epsilon)
            layer.b -= learning_rate * layer.b.grad / (torch.sqrt(layer.sb) + epsilon)
```

We now develop a feed-forward neural network with the architecture $784 \rightarrow 40(ReLU) \rightarrow 30(ReLU) \rightarrow 10(softmax)$. You can choose your own way to implement your network and an optimizer of interest. You should train model in $50$ epochs and evaluate the trained model on the test set.

```python
#Your code here
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

class FFN(torch.nn.Module):
  def __init__(self, input_size=28*28, n1=40, n2=30, num_classes=10, act=torch.nn.ReLU):
    super().__init__()
    self.net = torch.nn.Sequential(
        torch.nn.Linear(input_size, n1),
        act(),
        torch.nn.Linear(n1, n2),
        act(),
        torch.nn.Linear(n2, num_classes),
    )

  def forward(self, x):
    x = x.view(x.size(0), -1)
    return self.net(x)


def compute_loss(model, loss_fn, loader):
  model.eval()
  loss = 0.0
  with torch.no_grad():
    for batchX, batchY in loader:
      batchX, batchY = batchX.to(device).float(), batchY.to(device).long()
      loss += loss_fn(model(batchX), batchY).item()
  model.train()
  return loss / len(loader)


def compute_acc(model, loader):
  model.eval()
  correct, total = 0, 0
  with torch.no_grad():
    for batchX, batchY in loader:
      batchX, batchY = batchX.to(device).float(), batchY.to(device).long()
      preds = torch.argmax(model(batchX), dim=1)
      correct += (preds == batchY).sum().item()
      total += batchY.size(0)
  model.train()
  return correct / total


ffn_model = FFN(input_size=28*28, n1=40, n2=30, num_classes=10, act=torch.nn.ReLU).to(device)
loss_fn = torch.nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(ffn_model.parameters(), lr=1e-3)

num_epochs = 10
for epoch in range(num_epochs):
    for batchX, batchY in train_loader:
        batchX, batchY = batchX.to(device).float(), batchY.to(device).long()
        optimizer.zero_grad()
        loss = loss_fn(ffn_model(batchX), batchY)
        loss.backward()
        optimizer.step()
    train_loss = compute_loss(ffn_model, loss_fn, train_loader)
    train_acc = compute_acc(ffn_model, train_loader)
    val_loss = compute_loss(ffn_model, loss_fn, val_loader)
    val_acc = compute_acc(ffn_model, val_loader)
    print(f"Epoch {epoch+1}/{num_epochs}, Train Loss: {train_loss:.4f}, Train Acc: {train_acc*100:.2f}%, Val Loss: {val_loss:.4f}, Val Acc: {val_acc*100:.2f}%")

test_loss = compute_loss(ffn_model, loss_fn, test_loader)
test_acc = compute_acc(ffn_model, test_loader)
print(f"Test Loss: {test_loss:.4f}, Test Acc: {test_acc*100:.2f}%")
```

####  <font color="red">**Question 2.4:**</font> Implement the loss with the form: $loss(p,y)=CE(1_{y},p)+\lambda H(p)$ where $H(p)=-\sum_{i=1}^{M}p_{i}\log p_{i}$ is the entropy of $p$, $p$ is the prediction probabilities of a data point $x$ with the ground-truth label $y$, $1_y$ is an one-hot label, and $\lambda >0$ is a trade-off parameter. Set $\lambda = 0.1$ to train a model.

<div style="text-align: right"> <font color="red">[5 points]</font> </div>


```python
#Your code here
# Question 2.4:  loss(p, y) = CE(1_y, p) + lambda * H(p)   with   H(p) = -sum_i p_i log p_i

class EntropyRegularisedCE(torch.nn.Module):
    """
    loss(p, y) = CE(1_y, p) + lambda * H(p)

    p = softmax(h) are the prediction probabilities of a data point with logits h,
    1_y is the one-hot ground-truth label, so CE(1_y,p) = -log p_y, and
    H(p) = -sum_{i=1}^{M} p_i log p_i is the entropy of p.
    """
    def __init__(self, lam=0.1):
        super().__init__()
        self.lam = lam

    def forward(self, logits, y):
        log_p = torch.nn.functional.log_softmax(logits, dim=1)   # log p_i (numerically stable softmax + log)
        p = torch.nn.functional.softmax(logits, dim=1)
        ce = -log_p.gather(dim=1, index=y.unsqueeze(1)).squeeze(1)  # CE(1_y, p) = -log p_y  (one value per example)
        entropy = -(p * log_p).sum(dim=1)                          # H(p) = -sum_i p_i log p_i  (one value per example)
        return (ce + self.lam * entropy).mean()


def train_one_epoch(model, loss_fn, optimizer, loader):
    """One pass over the training set with a standard optimizer step."""
    model.train()
    for batchX, batchY in loader:
        batchX, batchY = batchX.to(device).float(), batchY.to(device).long()
        optimizer.zero_grad()
        loss = loss_fn(model(batchX), batchY)
        loss.backward()
        optimizer.step()


def fit_ffn(model, loss_fn, optimizer, num_epochs=10, train_step=train_one_epoch):
    """
    Train `model` with `loss_fn` and `optimizer` for `num_epochs` epochs and report the train/val
    metrics after each epoch. `train_step` is the function used to train one epoch
    (it is replaced by the SAM step in Question 2.5). Returns the history of the metrics.
    """
    ce_loss = torch.nn.CrossEntropyLoss()
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    for epoch in range(num_epochs):
        train_step(model, loss_fn, optimizer, train_loader)
        train_loss = compute_loss(model, ce_loss, train_loader)   # reported with the plain CE so that runs are comparable
        train_acc = compute_acc(model, train_loader)
        val_loss = compute_loss(model, ce_loss, val_loader)
        val_acc = compute_acc(model, val_loader)
        for k, v in zip(history, (train_loss, train_acc, val_loss, val_acc)):
            history[k].append(v)
        print(f"Epoch {epoch+1}/{num_epochs}, Train Loss: {train_loss:.4f}, Train Acc: {train_acc*100:.2f}%, "
              f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc*100:.2f}%")
    return history


# Train the 784 -> 40 (ReLU) -> 30 (ReLU) -> 10 network of Question 2.2 with the new loss (lambda = 0.1)
torch.manual_seed(1234)
lam = 0.1
entropy_loss_fn = EntropyRegularisedCE(lam=lam)
ffn_entropy = FFN(input_size=28*28, n1=40, n2=30, num_classes=10, act=torch.nn.ReLU).to(device)
optimizer = torch.optim.Adam(ffn_entropy.parameters(), lr=1e-3)

num_epochs = 10
history_entropy = fit_ffn(ffn_entropy, entropy_loss_fn, optimizer, num_epochs=num_epochs)

test_loss_entropy = compute_loss(ffn_entropy, torch.nn.CrossEntropyLoss(), test_loader)
test_acc_entropy = compute_acc(ffn_entropy, test_loader)
print(f"\n[CE + {lam} * H(p)]  Test CE Loss: {test_loss_entropy:.4f}, Test Acc: {test_acc_entropy*100:.2f}%")


```

```python
#Your code here

class SAM(torch.optim.Optimizer):
    """
    SAM wraps a base optimizer (e.g. Adam). Usage per mini-batch:
        loss_fn(model(x), y).backward();  sam.first_step(zero_grad=True)  steps 1-2
        loss_fn(model(x), y).backward();  sam.second_step(zero_grad=True) steps 3-4
    """
    def __init__(self, params, base_optimizer, rho=0.05, **kwargs):
        assert rho >= 0.0, "rho must be non-negative"
        defaults = dict(rho=rho, **kwargs)
        super().__init__(params, defaults)
        self.base_optimizer = base_optimizer(self.param_groups, **kwargs)
        self.param_groups = self.base_optimizer.param_groups   # share the parameter groups with the base optimizer

    @torch.no_grad()
    def first_step(self, zero_grad=False):
        """Climb to the adversarial point w + eps_hat (Eq. 2). The gradient of L_B(w) must have been computed."""
        grad_norm = self._grad_norm()
        for group in self.param_groups:
            scale = group["rho"] / (grad_norm + 1e-12)
            for p in group["params"]:
                if p.grad is None:
                    continue
                e_w = p.grad * scale                     # eps_hat = rho * g / ||g||
                p.add_(e_w)                               # w  ->  w + eps_hat
                self.state[p]["e_w"] = e_w
        if zero_grad:
            self.zero_grad()

    @torch.no_grad()
    def second_step(self, zero_grad=False):
        """Go back to w and update it with the gradient computed at w + eps_hat (Eq. 3)."""
        for group in self.param_groups:
            for p in group["params"]:
                if p.grad is None or "e_w" not in self.state[p]:
                    continue
                p.sub_(self.state[p]["e_w"])              # w + eps_hat  ->  w
        self.base_optimizer.step()                        # w <- w - eta * g_sam  (base optimizer update rule)
        if zero_grad:
            self.zero_grad()

    def _grad_norm(self):
        """||grad L_B(w)||_2 over all parameters."""
        norms = [p.grad.norm(p=2) for group in self.param_groups for p in group["params"] if p.grad is not None]
        return torch.norm(torch.stack(norms), p=2)

    def step(self, closure=None):
        raise RuntimeError("Use first_step() and second_step() instead of step().")


def sam_train_one_epoch(model, loss_fn, optimizer, loader):
    """One pass over the training set with the two-pass SAM update (Algorithm 1 of the paper)."""
    model.train()
    for batchX, batchY in loader:
        batchX, batchY = batchX.to(device).float(), batchY.to(device).long()
        # 1st pass: gradient at w, then climb to w + eps_hat
        loss_fn(model(batchX), batchY).backward()
        optimizer.first_step(zero_grad=True)
        # 2nd pass: gradient at w + eps_hat, then go back to w and update with the base optimizer
        loss_fn(model(batchX), batchY).backward()
        optimizer.second_step(zero_grad=True)


# Apply SAM to the best architecture found in Question 2.3
best_n1, best_n2, best_act_name = best['n1'], best['n2'], best['act']
best_act = grid_act[best_act_name]
print(f"Best architecture from Q2.3: 784 -> {best_n1} ({best_act_name}) -> {best_n2} ({best_act_name}) -> 10\n")

num_epochs = 50
rho = 0.05                     # neighbourhood size (the value used for most experiments in the paper)
loss_fn = torch.nn.CrossEntropyLoss()

# (i) best architecture trained with SAM (base optimizer: Adam, lr = 1e-3, as in Q2.3)
torch.manual_seed(1234)
ffn_sam = FFN(input_size=28*28, n1=best_n1, n2=best_n2, num_classes=10, act=best_act).to(device)
sam_optimizer = SAM(ffn_sam.parameters(), torch.optim.Adam, rho=rho, lr=1e-3)
print("Training with SAM:")
history_sam = fit_ffn(ffn_sam, loss_fn, sam_optimizer, num_epochs=num_epochs, train_step=sam_train_one_epoch)

# (ii) the same architecture trained for the same number of epochs WITHOUT SAM (plain Adam) as the baseline
torch.manual_seed(1234)
ffn_base = FFN(input_size=28*28, n1=best_n1, n2=best_n2, num_classes=10, act=best_act).to(device)
base_optimizer = torch.optim.Adam(ffn_base.parameters(), lr=1e-3)
print("\nTraining without SAM (baseline):")
history_base = fit_ffn(ffn_base, loss_fn, base_optimizer, num_epochs=num_epochs)

print("\nEvaluation on the test set")
for name, m in [("Adam + SAM", ffn_sam), ("Adam (no SAM)", ffn_base)]:
    print(f"{name:14s}: Test Loss: {compute_loss(m, loss_fn, test_loader):.4f}, Test Acc: {compute_acc(m, test_loader)*100:.2f}%")
```

```python
#Your code here
class YourBlock(nn.Module):
  def __init__(self, in_feature_maps, out_feature_maps, drop_rate = 0.2, batch_norm = True, use_skip = True):
    super(YourBlock, self).__init__()
    self.use_skip = use_skip
    #Your code here
    # Block pattern: [conv, batch norm, activation, conv, batch norm, activation, max pool, dropout]
    # conv: 3x3 filters, stride 1, padding 1 (keeps the spatial size); max pool: kernel 2, stride 2, padding 0 (halves it)
    # If batch_norm is False the BatchNorm2d layers are replaced by Identity (no-op) layers.
    bn = (lambda c: nn.BatchNorm2d(c)) if batch_norm else (lambda c: nn.Identity())
    self.block = nn.ModuleList([
        nn.Conv2d(in_feature_maps, out_feature_maps, kernel_size=(3, 3), stride=(1, 1), padding=1),
        bn(out_feature_maps),
        nn.ReLU(inplace=True),
        nn.Conv2d(out_feature_maps, out_feature_maps, kernel_size=(3, 3), stride=(1, 1), padding=1),
        bn(out_feature_maps),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(kernel_size=2, stride=2, padding=0),
        nn.Dropout(p=drop_rate),
    ])
    # Skip connection: a 1x1 convolution with stride 2 maps the block input [in, H, W] to the
    # shape of the block output [out, H/2, W/2] so that the two can be added.
    if self.use_skip:
        self.skip_conv = nn.Conv2d(in_feature_maps, out_feature_maps, kernel_size=(1, 1), stride=(2, 2))

  def forward(self, x):
    #Write your code here
    out = x
    for layer in self.block:
        out = layer(out)
    if self.use_skip:
        out = out + self.skip_conv(x)      # residual connection: block(x) + skip(x)
    return out
```

**Second, you need to use the above `YourBlock` to implement the class `YourCNN`**.

```python
class YourCNN(nn.Module):
  def __init__(self, list_feature_maps = [16, 32, 64], drop_rate = 0.2, batch_norm= True, use_skip = True, num_classes = 20):
    super(YourCNN, self).__init__()
    layers = []

    # Stem: one 3x3 conv (3 -> list_feature_maps[0]) with batch norm and ReLU, as in the model configuration above
    layers += [nn.Conv2d(3, list_feature_maps[0], kernel_size=(3, 3), stride=(1, 1), padding='same'),
               nn.BatchNorm2d(list_feature_maps[0]) if batch_norm else nn.Identity(),
               nn.ReLU(inplace=True)]
    # One YourBlock per pair of consecutive feature-map sizes: [16, 32, 64] -> blocks 16 -> 32 and 32 -> 64
    for in_maps, out_maps in zip(list_feature_maps[:-1], list_feature_maps[1:]):
        layers.append(YourBlock(in_maps, out_maps, drop_rate=drop_rate, batch_norm=batch_norm, use_skip=use_skip))
    # Global average pooling flattens the [C, H, W] tensor to a C-dimensional vector,
    # followed by the output linear layer that produces the logits
    layers += [nn.AdaptiveAvgPool2d(output_size=(1, 1)),
               nn.Flatten(start_dim=1, end_dim=-1),
               nn.Linear(in_features=list_feature_maps[-1], out_features=num_classes, bias=True)]
    self.block = nn.ModuleList(layers)
    self.num_features = list_feature_maps[-1]     # size of the penultimate layer (used in Question 3.5)

  def features(self, x):
    """Output of the penultimate layer (the flattened global-average-pooled feature vector)."""
    for layer in self.block[:-1]:
        x = layer(x)
    return x

  def forward(self, x):
    #Write your code here
    return self.block[-1](self.features(x))     # logits
```

```python
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

def compute_loss(model, loss_fn, loader):
  loss = 0
  # Set model to eval mode for inference
  model.eval()
  with torch.no_grad():  # No need to track gradients for validation
    for (batchX, batchY) in loader:
      # Move data to the same device as the model
      batchX, batchY = batchX.to(device).type(torch.float32), batchY.to(device).type(torch.long)
      loss += loss_fn(model(batchX), batchY)
  # Set model back to train mode
  model.train()
  return float(loss)/len(loader)
```

```python
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

def compute_acc(model, loader):
    correct = 0
    totals = 0
    # Set model to eval mode for inference
    model.eval()
    for (batchX, batchY) in loader:
        # Move batchX and batchY to the same device as the model
        batchX, batchY = batchX.to(device).type(torch.float32), batchY.to(device)
        outputs = model(batchX)  # feed batch to the model
        totals += batchY.size(0)  # accumulate totals with the current batch size
        predicted = torch.argmax(outputs.data, 1)  # get the predicted class
        # Move batchY to the same device as predicted for comparison
        correct += (predicted == batchY).sum().item()
    return correct / totals
```

#### <font color="red">**Question 3.4:**</font> **Implement the **one-versus-all (OVA)** loss**

The details are as follows:
- You need to apply `the sigmoid activation function` to logits $h = [h_1, h_2,...,h_M]$ instead of `the softmax activation` function as usual to obtain $p = [p_1, p_2,...,p_M]$, meaning that $p_i = sigmoid(h_i), i=1,...,M$. Note that $M$ is the number of classes.
- Given a data example $x$ with the ground-truth label $y$, the idea is to maximize the likelihood $p_y$ and to minimize the likelihoods $p_i, i \neq y$. Therefore, the objective function is to find the model parameters to
  - $\max\left\{ \log p_{y}+\sum_{i\neq y}\log(1-p_{i})\right\}$ or equivalently $\min\left\{ -\log p_{y}-\sum_{i\neq y}\log(1-p_{i})\right\}$.
  - For example, if $M=3$ and $y=2$, you need to minimize $\min\left\{ -\log(1-p_{1})-\log p_{2}-\log(1-p_{3})\right\}$.

Compare the model trained with the OVA loss and the same model trained with the standard cross-entropy loss.

<div style="text-align: right"> <font color="red">[4 points]</font> </div>



```python
#Your code here


def ova_loss(logits, y):
    """
    One-versus-all loss:  loss(h, y) = -log p_y - sum_{i != y} log(1 - p_i)   with   p_i = sigmoid(h_i)

    logits h has shape [batch_size, M]; each class i gets its own binary classifier "class i vs the rest",
    so we push p_y towards 1 and every other p_i towards 0. Averaged over the mini-batch.
    """
    log_p = torch.nn.functional.logsigmoid(logits)            # log p_i = log sigmoid(h_i)          (numerically stable)
    log_1_minus_p = torch.nn.functional.logsigmoid(-logits)   # log(1 - p_i) = log sigmoid(-h_i)
    one_hot = torch.nn.functional.one_hot(y, num_classes=logits.size(1)).float()   # 1 at position y, 0 elsewhere
    per_example = -(one_hot * log_p + (1 - one_hot) * log_1_minus_p).sum(dim=1)     # -log p_y - sum_{i!=y} log(1-p_i)
    return per_example.mean()

def ova_batch_loss(model, X, y):
    return ova_loss(model(X), y)

# Train the same CNN with the OVA loss; the prediction is still argmax_i p_i = argmax_i h_i
my_cnn_ova = YourCNN(list_feature_maps=[16, 32, 64], use_skip=True)
history_ova = fit_custom(my_cnn_ova, train_loader, val_loader, ova_batch_loss, num_epochs=10)


print(summarize("OVA loss", history_ova))
print(f"OVA loss of the OVA model on the validation set: {compute_loss(my_cnn_ova, ova_loss, val_loader):.4f}")
```

#### <font color="red">**Question 3.5:**</font> **Incorporate the unsupervised loss to the cross-entropy loss**

The unsupervised loss is defined as follows
- For each mini-batch $x$ of images, we rotate these images by $0^{\circ}, 90^{\circ}, 180^{\circ}, 270^{\circ}$ and then predict for labels $0,1,2,3$ respectively. For example, if we rotate an image by $180^{\circ}$, we need to predict the ground-truth label $2$.
- In terms of architecture, on the top of the penultimate layer (i.e., the layer right before the current output layer) of your CNN, you add one more output layer to predict the rotation angles ($0^{\circ}$ - label 0, $90^{\circ}$ - label 1, $180^{\circ}$ - label 2, and $270^{\circ}$ - label 3).
- The final loss is $total\_loss=CE\_loss+\lambda\times unsupervised\_loss$ where $\lambda > 0$ is a trade-off parameter.
-  You can set $\lambda$ (e.g., $\lambda = 0.1$) a value and train a model with this value.
-  Note that you can as many cells as you want to extend the current code.
   
<div style="text-align: right"> <font color="red">[6 points]</font> </div>

```python
#Your code here


class RotationCNN(YourCNN):
    """YourCNN with an extra output layer that predicts the rotation angle (4 classes)."""
    def __init__(self, list_feature_maps=[16, 32, 64], drop_rate=0.2, batch_norm=True, use_skip=True, num_classes=20):
        super().__init__(list_feature_maps, drop_rate, batch_norm, use_skip, num_classes)
        self.rotation_head = nn.Linear(self.num_features, 4)

    def forward_rotation(self, x):
        return self.rotation_head(self.features(x))      # rotation logits from the penultimate layer

def rotate_batch(X):
    """Returns the 4 rotated copies of X (0, 90, 180, 270 degrees) and their rotation labels 0, 1, 2, 3."""
    X_rot = torch.cat([torch.rot90(X, k, dims=(2, 3)) for k in range(4)], dim=0)
    r = torch.arange(4, device=X.device).repeat_interleave(X.size(0))
    return X_rot, r

lambda_rot = 0.1

def rotation_batch_loss(model, X, y, lam=lambda_rot):
    ce_loss = loss_fn(model(X), y)                       # supervised loss on the original images
    X_rot, r = rotate_batch(X)
    unsupervised_loss = loss_fn(model.forward_rotation(X_rot), r)   # rotation prediction on the rotated images
    return ce_loss + lam * unsupervised_loss

# Show the four rotations of the first two images of a batch (labels shown are the class labels)
X_rot_vis, r_vis = rotate_batch(X_vis[:2])
print("Rotations 0, 90, 180, 270 degrees of two training images (rotation labels:", r_vis.tolist(), ")")
visualize_data(X_rot_vis.numpy(), y_vis[:2].repeat(4).numpy(), images_per_row=8)

my_cnn_rot = RotationCNN(list_feature_maps=[16, 32, 64], use_skip=True)
history_rot = fit_custom(my_cnn_rot, train_loader, val_loader, rotation_batch_loss, num_epochs=10)

# accuracy of the auxiliary rotation task on the validation set
@torch.no_grad()
def rotation_acc(model, loader):
    model.eval(); correct, total = 0, 0
    for X, _ in loader:
        X_rot, r = rotate_batch(X.to(device).type(torch.float32))
        correct += (model.forward_rotation(X_rot).argmax(1) == r).sum().item(); total += r.numel()
    return correct / total

print()
print(summarize("Baseline CE (Q3.1)", history))
print(summarize("Mixup", history_mixup))
print(summarize("CutMix", history_cutmix))
print(summarize("OVA loss", history_ova))
print(summarize(f"CE + {lambda_rot} * rotation loss", history_rot))
print(f"Rotation-prediction accuracy of the auxiliary head on the validation set: {rotation_acc(my_cnn_rot, val_loader)*100:.2f}%")
```
